**Python 里 `int` 的内存占用不是固定的**。

`bit_length()` 返回的是“表示这个整数最少需要多少二进制位”，比如：
```python
a = 10
a.bit_length()   # 4，因为 10 的二进制是 1010
```
它并不是直接告诉你这个对象占了多少字节，但数值越大，需要的二进制位越多，底层存储也会随之变大。

### 为什么 Python 整数内存会变？
CPython 里的 `int` 是**任意精度整数**，不是 C 语言那种固定 32 位或 64 位的整数。
它的底层大致长这样：
```c
struct PyLongObject {
    PyObject_VAR_HEAD   // 引用计数、类型指针、长度信息等
    digit ob_digit[];   // 动态数组，用来存整数值
};
```
整数越大，`ob_digit` 数组就越长，所以内存会动态增加。
在 64 位 CPython 中：

- 一个普通小 `int` 对象大约 **28 字节**左右；
- 数值超过 `2^30` 后，每多一个“数字块”，大约多占 **4 字节**；
- 所以 `0`、`1`、`100000` 占用的内存并不完全一样。

### 还有一个小整数缓存
Python 还会把 `-5` 到 `256` 这些常用整数提前创建好并复用，所以这些小整数不仅内存固定，而且多次赋值可能指向同一个对象。
所以可以这样理解：
- `bit_length()`：表示数值本身需要多少二进制位；
- `int` 对象内存：包括对象头 + 数值存储部分；
- 数值越大，`int` 对象占用的内存越多。

## 正则表达式
正则表达式一般用来解析字符串对象
tokenizer也经常会用到这个东西

````python
import re
series = """
        '01/18/2014 13:00:00', 100, '1st'; 
        '01/18/2014 13:30:00', 110, '2nd'; 
        '01/18/2014 14:00:00', 120, '3rd' 
        """
dt = re.compile("'[0-9/:\\s]+'")  #这里编译器提醒我要用\\s 而不是\s
result = dt.findall(series)
print(result)
````

````python
#列表是可变的
l = [1, 2.5, 'data']
l.append([4,3])
print("append:", l) #加到最后面
l.extend([1.0, 1.5, 2.0])
print("extend:",l)
l.insert(1,'insert') #插入最前面
print("insert:",l)
l.remove('data')
print("remove:",l)

p = l.pop(3)
print("pop:",l,p)
#从l中删除index为3的对象，并返回该对象

print(l[2:5]) #切片
````
