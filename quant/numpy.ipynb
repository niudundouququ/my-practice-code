为了避免这种现象，可以用copy库的deepcopy函数
即用物理副本代替引用指针

```python
from copy import deepcopy 
v = [0.5, 0.75, 1.0, 1.5, 2.0] 
m = 3 * [deepcopy(v), ] 

v[0] = "牛顿斗蛐蛐"
print(m)
```

### NumPy 结构数组
用户可自行定义数组的结构，如下是个例子

```python
dt = np.dtype([('Name','S10'),('Age','i4'),('Height','f'),('Children/Pets','i4',2)])
```

这里是在定义数据类型（dtype）。就像在数据库建表时，你要规定每一列叫什么名字，以及存什么格式的数据。

('Name', 'S10')：定义了一个叫 Name 的列，S10 表示这是一个字符串，最长占用 10 个字节。

('Age', 'i4')：定义了一个叫 Age 的列，i4 表示这是一个 4字节的整数（即 32位整数）。

('Height', 'f')：定义了一个叫 Height 的列，f 表示这是一个浮点数（小数）。

('Children/Pets', 'i4', 2)：这一列比较特殊。它叫 Children/Pets，类型是整数，但最后的 2 表示这是一个包含 2 个整数的子数组。比如你可以存 (孩子数量, 宠物数量)。

```python
#等价的写法是：
dt = np.dtype({'names': ['Name', 'Age', 'Height', 'Children/Pets'],  'formats':'O int float int,int'.split()}) 
#运用了split函数把format分割出来
```

```python
s = np.array([('Smith',45,1.83,(2,1)),('Peter',18,180,(0,0))],dtype=dt)
s
```
