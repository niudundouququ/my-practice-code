## 用NumPy进行数值计算

### 本节先使用python

```python
v = [0.5, 0.75, 1.0, 1.5, 2.0]
m = [v, v, v]
print(m)
n = [m, m, m]
print(n)
print("但是这里请注意，用v，m构建的更高级数组在操作的时候，实际上操作的是v，m本身，也就是说根本没有复制v，m只是用指针指向他们\n例：")

v[0] = "牛顿斗蛐蛐"
print(m,'\n',n)
```

为了避免这种现象，可以用copy库的deepcopy函数
即用物理副本代替引用指针

```python
from copy import deepcopy 
v = [0.5, 0.75, 1.0, 1.5, 2.0] 
m = 3 * [deepcopy(v), ] 

v[0] = "牛顿斗蛐蛐"
print(m)
```

数组：array 与cpp的数组完全一样，只能存储单一类型的元素

```python
v = [0.5, 0.75, 1.0, 1.5, 2.0] 
import array

a = array.array('f',v) #'f'决定了该数组存储变量的类型
a.append(0.5)
a.extend([114.5, 14.1]) #之所以输出14.10000381469727,是因为f单精度不够，换成double就行了
a
```
